## 01. WHAT IS SQL?

**Definition:** SQL stands for Structured Query Language. It is a standard language used to communicate with relational database management systems.

**Uses of SQL:**
1. Create databases and tables.
2. Insert, update and delete records.
3. Retrieve information using queries.
4. Define relationships and constraints.
5. Manage transactions and permissions.

**Example:** A college can use SQL to retrieve student names, update marks or find students enrolled in a particular course.

## 02. WHAT IS A DATABASE?

**Definition:** A database is an organized collection of related data that can be stored, accessed, managed and updated.

**Example:** A college database may contain student, teacher, course, department and examination information.

**Why databases are used:**
- Store information systematically.
- Search and retrieve data efficiently.
- Update and manage records.
- Reduce unnecessary duplication.
- Support controlled access and data consistency.

**Common database types:** Relational, document, key-value, graph and time-series databases.

```python
CREATE DATABASE college;

USE college;

SHOW DATABASES;
```

## Explanation

## 03. WHAT IS DBMS?

**Definition:** DBMS stands for Database Management System. It is software used to create, store, retrieve, update and manage data in databases.

**Examples:** MySQL, PostgreSQL, Oracle Database, Microsoft SQL Server and SQLite.

**Main functions:**
1. Data storage and retrieval.
2. Insertion, updating and deletion.
3. User access control.
4. Backup and recovery.
5. Data integrity management.
6. Concurrency control.

**Example:** A library uses a DBMS to manage books, members, loans and return dates.

```python
CREATE TABLE books (
    book_id INT PRIMARY KEY,
    title VARCHAR(100),
    author VARCHAR(100)
);

INSERT INTO books (book_id, title, author)
VALUES (1, 'Database Systems', 'R. Author');

SELECT * FROM books;
```

## Expected Output

| book_id | title | author |
|---:|---|---|
| 1 | Database Systems | R. Author |

The DBMS stores the table and record, then returns the record through `SELECT`.

## 04. WHAT IS RDBMS?

**Definition:** RDBMS stands for Relational Database Management System. It is a type of DBMS that stores data in related tables consisting of rows and columns.

**Important concepts:**
- **Table:** Stores related data.
- **Row / Record:** One individual entry.
- **Column / Field:** One attribute.
- **Primary Key:** Uniquely identifies a row.
- **Foreign Key:** Refers to a key in another table to establish a relationship.

**Examples:** MySQL, PostgreSQL, Oracle Database and Microsoft SQL Server.

```python
CREATE TABLE courses (
    course_id INT PRIMARY KEY,
    course_name VARCHAR(100)
);

CREATE TABLE students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(100),
    course_id INT,
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
);
```

## Explanation

The `course_id` foreign key in `students` refers to `courses.course_id`, connecting student records to course records.

## 05. DIFFERENCE BETWEEN DBMS AND RDBMS

**DBMS:** General software for managing databases. It may use different data models.

**RDBMS:** A type of DBMS that organizes data into related tables and commonly uses keys and constraints.

| Feature | DBMS | RDBMS |
|---|---|---|
| Meaning | Database Management System | Relational Database Management System |
| Data model | Can use different models | Relational model |
| Structure | Depends on the system | Tables with rows and columns |
| Relationships | Depends on the model | Supported through keys |
| Integrity rules | Depends on the system | Commonly enforced with constraints |
| Examples | Database systems in general | MySQL, PostgreSQL, Oracle |

**Important:** Every RDBMS is a DBMS, but not every DBMS is an RDBMS.

```python
CREATE TABLE departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(50)
);

CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    employee_name VARCHAR(100),
    department_id INT,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
);
```

## 07. FEATURES OF SQL

1. **Data Definition:** Create, alter and remove database structures.
2. **Data Manipulation:** Insert, update and delete records.
3. **Data Retrieval:** Fetch selected information.
4. **Filtering and Sorting:** Filter rows and sort results.
5. **Joins:** Retrieve related data from multiple tables.
6. **Constraints:** Enforce rules such as primary and foreign keys.
7. **Aggregate Functions:** Calculate count, sum and average.
8. **Transactions:** Commit or roll back groups of changes.

Exact features and syntax can vary between database systems.

```python
SELECT department_id, COUNT(*) AS employee_count
FROM employees
WHERE department_id IS NOT NULL
GROUP BY department_id
ORDER BY employee_count DESC;
```

## Explanation

This query filters out rows with a missing department, groups employees by department, counts each group and sorts by the count in descending order. It assumes an `employees` table with a `department_id` column.

## 09. WHAT ARE SQL COMMANDS?

**Definition:** SQL commands are instructions written in SQL to perform operations on a database.

They are used to:
- Define database structures.
- Insert, modify and delete data.
- Retrieve data.
- Control permissions.
- Manage transactions.

Examples include `CREATE TABLE`, `INSERT`, `SELECT`, `UPDATE` and `DELETE`.

```python
CREATE TABLE sample_students (
    id INT PRIMARY KEY,
    name VARCHAR(50)
);

INSERT INTO sample_students (id, name)
VALUES (1, 'Ankan');

SELECT * FROM sample_students;

UPDATE sample_students
SET name = 'Ankan Haldar'
WHERE id = 1;

DELETE FROM sample_students
WHERE id = 1;
```

## Explanation

The statements create a table, insert a row, retrieve it, update it and finally delete it. Run them in order.

## 10. TYPES OF SQL COMMANDS

SQL commands are commonly grouped into five categories:

1. **DDL – Data Definition Language:** Defines or changes database structures.
2. **DML – Data Manipulation Language:** Inserts, updates and deletes data.
3. **DQL – Data Query Language:** Retrieves data.
4. **DCL – Data Control Language:** Manages access permissions.
5. **TCL – Transaction Control Language:** Manages transactions.

Each category is explained separately in the following topics.

## 11. DDL – DATA DEFINITION LANGUAGE

**Definition:** DDL is used to create, change and remove database objects such as databases, tables and indexes.

**Common MySQL DDL commands:**
- `CREATE` – creates a database object.
- `ALTER` – changes an object's structure.
- `DROP` – removes a database object.
- `TRUNCATE` – removes all rows while keeping the table structure.
- `RENAME TABLE` – changes a table's name.

**Example use:** Creating and modifying a student table.

```python
CREATE TABLE ddl_students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(100)
);

ALTER TABLE ddl_students
ADD email VARCHAR(100);

RENAME TABLE ddl_students TO students_ddl_demo;

TRUNCATE TABLE students_ddl_demo;

DROP TABLE students_ddl_demo;
```

## Explanation

The statements create a table, add a column, rename it, remove all rows and finally remove the table. `DROP TABLE` is destructive; use it only on practice tables.

## 12. DML – DATA MANIPULATION LANGUAGE

**Definition:** DML is used to add, modify and remove records stored in database tables.

**Common DML commands:**
- `INSERT` – adds new records.
- `UPDATE` – modifies existing records.
- `DELETE` – removes selected records.

Some textbooks classify `SELECT` separately as DQL; broader classifications may include it under DML.

```python
CREATE TABLE dml_students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(100),
    marks INT
);

INSERT INTO dml_students (student_id, student_name, marks)
VALUES (1, 'Ankan', 85);

UPDATE dml_students
SET marks = 90
WHERE student_id = 1;

DELETE FROM dml_students
WHERE student_id = 1;
```

## Explanation

The example inserts a student, updates the student's marks and deletes that row. The `WHERE` clause is important in `UPDATE` and `DELETE` because it selects the affected rows.

## 13. DQL – DATA QUERY LANGUAGE

**Definition:** DQL is used to retrieve data from database tables.

**Main command:** `SELECT`.

It can be used to:
- Select particular columns or all columns.
- Filter rows using `WHERE`.
- Sort results using `ORDER BY`.
- Group rows using `GROUP BY`.

**Example use:** Displaying students who scored above 80 marks.

```python
SELECT student_id, student_name, marks
FROM students
WHERE marks > 80
ORDER BY marks DESC;
```

## Explanation

The query selects three columns, keeps rows where marks are greater than 80 and sorts the result from highest to lowest marks. It assumes the `students` table contains these columns.

## 14. DCL – DATA CONTROL LANGUAGE

**Definition:** DCL is used to manage permissions and access privileges for database users.

**Common DCL commands:**
- `GRANT` – gives specified privileges to a user.
- `REVOKE` – removes specified privileges from a user.

**Why DCL is important:** It controls who can access database objects and limits users to the operations they need.

```python
-- Practice only; administrator privileges may be required.

CREATE USER 'practice_user'@'localhost'
IDENTIFIED BY 'Use_A_Strong_Password_Here';

GRANT SELECT, INSERT
ON college.*
TO 'practice_user'@'localhost';

REVOKE INSERT
ON college.*
FROM 'practice_user'@'localhost';
```

## Explanation and Safety

`CREATE USER` creates an account. `GRANT` gives it SELECT and INSERT privileges on the `college` database. `REVOKE` removes INSERT permission. Use a strong password and never publish real credentials in GitHub.

## 15. TCL – TRANSACTION CONTROL LANGUAGE

**Definition:** TCL is used to manage transactions in a database.

**Common TCL commands:**
- `COMMIT` – permanently saves changes made in the current transaction.
- `ROLLBACK` – cancels uncommitted changes.
- `SAVEPOINT` – creates a point within a transaction.
- `ROLLBACK TO SAVEPOINT` – rolls back changes made after a savepoint.

**Example use:** A bank transfer should subtract money from one account and add it to another as one transaction.

**Important:** Transaction behaviour depends on the storage engine and configuration. In MySQL, use a transactional engine such as InnoDB.

```python
START TRANSACTION;

UPDATE accounts
SET balance = balance - 500
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 500
WHERE account_id = 2;

COMMIT;
```

## Example: SAVEPOINT and ROLLBACK

This is a separate practice example. It assumes an `accounts` table and should be run independently.

```python
START TRANSACTION;

UPDATE accounts
SET balance = balance - 100
WHERE account_id = 1;

SAVEPOINT after_first_update;

UPDATE accounts
SET balance = balance + 100
WHERE account_id = 2;

ROLLBACK TO SAVEPOINT after_first_update;

ROLLBACK;
```

## Explanation

The first example transfers 500 units and commits both updates. In the second example, the transaction rolls back to the savepoint and then rolls back the remaining transaction, leaving no uncommitted changes from that transaction.

## 16. SQL DATA TYPES

**Definition:** SQL data types specify what kind of value a column can store. Choosing suitable data types helps maintain valid data and use storage efficiently.

### A. Numeric Data Types
- `INT` – whole numbers.
- `SMALLINT` – smaller-range whole numbers.
- `BIGINT` – large-range whole numbers.
- `DECIMAL(p, s)` – exact fixed-point numbers; useful for money.
- `FLOAT` – approximate floating-point numbers.
- `DOUBLE` – approximate floating-point numbers with greater precision than FLOAT.

### B. String Data Types
- `CHAR(n)` – fixed-length character string.
- `VARCHAR(n)` – variable-length character string.
- `TEXT` – longer text data.
- `TINYTEXT`, `MEDIUMTEXT`, `LONGTEXT` – text types with different capacities.

### C. Date and Time Data Types
- `DATE` – date value.
- `TIME` – time value.
- `DATETIME` – date and time value.
- `TIMESTAMP` – timestamp value with MySQL-specific range and time-zone behaviour.
- `YEAR` – year value.

### D. Binary Data Types
- `BINARY` and `VARBINARY` – binary data.
- `BLOB`, `TINYBLOB`, `MEDIUMBLOB`, `LONGBLOB` – binary large objects.

**Tip:** Use `DECIMAL` rather than `FLOAT` or `DOUBLE` when exact monetary values are required.

```python
CREATE TABLE data_type_demo (
    id INT,
    age SMALLINT,
    account_number BIGINT,
    price DECIMAL(10, 2),
    rating FLOAT,
    full_name VARCHAR(100),
    country_code CHAR(2),
    description TEXT,
    birth_date DATE,
    login_time TIME,
    created_at DATETIME,
    updated_at TIMESTAMP,
    profile_image BLOB
);
```

## Explanation

This table demonstrates common MySQL data types. Real table design should choose types and sizes based on application requirements.

## 17. SQL SYNTAX AND COMMENTS

**Definition:** SQL syntax is the set of rules used to write valid SQL statements.

### Basic SQL syntax rules
1. SQL keywords are conventionally written in uppercase for readability, although MySQL keywords are generally case-insensitive.
2. A semicolon (`;`) commonly terminates a statement.
3. Text values are usually enclosed in single quotes.
4. Numeric values are generally written without quotes.
5. Table and column names must follow valid identifier rules.
6. Indentation and line breaks make long queries easier to read.

- `-- ` – single-line comment (two hyphens followed by whitespace).
- `#` – single-line comment.
- `/* ... */` – multi-line comment.

Comments are notes and are not executed as SQL statements.
