## Topic 46: Arithmetic Operators

Arithmetic operators perform mathematical calculations on numeric values.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `salary + bonus` |
| `-` | Subtraction | `salary - 1000` |
| `*` | Multiplication | `salary * 12` |
| `/` | Division | `salary / 12` |
| `%` | Modulus (remainder) | `salary % 1000` |

Arithmetic expressions can be used in `SELECT` statements and in other SQL clauses where expressions are accepted.

### SQL Example — Topic 46

```sql
SELECT
    employee_name,
    salary,
    salary + 2000 AS salary_after_increment,
    salary - 1000 AS salary_after_deduction,
    salary * 12 AS annual_salary,
    salary / 12 AS monthly_average,
    salary % 1000 AS remainder_value
FROM operator_demo_employee;
```

**Expected result / explanation**

Each calculated column is an expression result. `AS` assigns a readable output name to the calculated column.

## Topic 47: Comparison Operators

Comparison operators compare two values and return a Boolean result used to filter or evaluate data.

| Operator | Meaning |
|---|---|
| `=` | Equal to |
| `!=` or `<>` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

They are commonly used in the `WHERE` clause. Comparisons involving `NULL` require `IS NULL` or `IS NOT NULL` rather than `=`.

### SQL Example — Topic 47

```sql
SELECT employee_name, salary
FROM operator_demo_employee
WHERE salary >= 45000;

SELECT employee_name, department
FROM operator_demo_employee
WHERE department <> 'IT';
```

**Expected result / explanation**

The first query returns employees whose salary is at least 45,000. The second excludes employees in the IT department.

## Topic 48: Logical Operators

Logical operators combine or reverse conditions.

- `AND` — all conditions must be true.
- `OR` — at least one condition must be true.
- `NOT` — reverses a condition.

Use parentheses when combining several conditions so the intended evaluation order is clear. SQL uses three-valued logic: a condition can be `TRUE`, `FALSE`, or `UNKNOWN` (often because of `NULL`).

### SQL Example — Topic 48

```sql
SELECT employee_name, department, salary
FROM operator_demo_employee
WHERE department = 'IT'
  AND salary > 50000;

SELECT employee_name, city
FROM operator_demo_employee
WHERE city = 'Kolkata'
   OR city = 'Howrah';

SELECT employee_name, department
FROM operator_demo_employee
WHERE NOT department = 'HR';
```

**Expected result / explanation**

The three queries demonstrate combined conditions, alternative conditions, and negation.

## Topic 49: AND Operator

`AND` returns true only when every condition is true. It is useful when a row must satisfy multiple requirements at the same time.

Truth table:
- TRUE AND TRUE = TRUE
- TRUE AND FALSE = FALSE
- FALSE AND FALSE = FALSE

### SQL Example — Topic 49

```sql
SELECT employee_name, department, salary, age
FROM operator_demo_employee
WHERE department = 'IT'
  AND salary > 50000
  AND age >= 25;
```

**Expected result / explanation**

Only employees satisfying all three conditions are returned.

## Topic 50: OR Operator

`OR` returns true when at least one condition is true.

Truth table:
- TRUE OR TRUE = TRUE
- TRUE OR FALSE = TRUE
- FALSE OR FALSE = FALSE

Use parentheses to group OR conditions when mixing them with AND.

### SQL Example — Topic 50

```sql
SELECT employee_name, department, city
FROM operator_demo_employee
WHERE department = 'HR'
   OR city = 'Durgapur';

SELECT employee_name, department, salary
FROM operator_demo_employee
WHERE (department = 'IT' OR department = 'Finance')
  AND salary >= 50000;
```

**Expected result / explanation**

The first query returns rows matching either condition. The second groups the department alternatives before applying the salary requirement.

## Topic 51: NOT Operator

`NOT` reverses a condition's truth value. It can be used with comparisons and predicates such as `IN`, `BETWEEN`, `LIKE`, and `EXISTS`.

Remember that `NOT` applied to an `UNKNOWN` result remains `UNKNOWN`; rows with `NULL` values are not automatically included.

### SQL Example — Topic 51

```sql
SELECT employee_name, department
FROM operator_demo_employee
WHERE NOT (department = 'IT');

SELECT employee_name, salary
FROM operator_demo_employee
WHERE NOT (salary < 40000);
```

**Expected result / explanation**

The first query excludes IT employees. The second returns rows where the salary is not less than 40,000.

## Topic 52: IN Operator

`IN` checks whether a value matches any value in a specified list or the result of a subquery. It is a shorter alternative to several equality comparisons joined by `OR`.

`NOT IN` checks that a value does not match any item in the list. Be careful when a `NOT IN` subquery can return `NULL`, because that can make the result `UNKNOWN`.

### SQL Example — Topic 52

```sql
SELECT employee_name, department
FROM operator_demo_employee
WHERE department IN ('IT', 'HR');

SELECT employee_name, city
FROM operator_demo_employee
WHERE city NOT IN ('Kolkata', 'Howrah');
```

**Expected result / explanation**

The first query returns IT and HR employees. The second excludes the two listed cities.

## Topic 53: BETWEEN Operator

`BETWEEN` checks whether a value falls within an inclusive range. Both endpoints are included.

`NOT BETWEEN` checks values outside the range. The range endpoints should be written in ascending order for readability.

### SQL Example — Topic 53

```sql
SELECT employee_name, salary
FROM operator_demo_employee
WHERE salary BETWEEN 40000 AND 55000;

SELECT employee_name, age
FROM operator_demo_employee
WHERE age NOT BETWEEN 24 AND 28;
```

**Expected result / explanation**

The first query includes salaries equal to 40,000 or 55,000. The second returns ages below 24 or above 28.

## Topic 54: LIKE Operator

`LIKE` performs pattern matching on character strings.

- `'A%'` — starts with A.
- `'%a'` — ends with a.
- `'%it%'` — contains `it`.
- `'_i%'` — has `i` as the second character.

In MySQL, case sensitivity depends on the character set and collation. Many common default collations are case-insensitive.

### SQL Example — Topic 54

```sql
SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE 'A%';

SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE '%a';

SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE '%iy%';
```

**Expected result / explanation**

Each query finds names matching the specified pattern. `%` and `_` are wildcard characters explained in the next topic.

## Topic 55: SQL Wildcards

Wildcards are special characters used with `LIKE` to represent unknown characters.

| Wildcard | Meaning | Example |
|---|---|---|
| `%` | Zero or more characters | `A%` |
| `_` | Exactly one character | `_i%` |

To search for a literal `%` or `_`, escape it. MySQL commonly supports backslash escaping, but the `ESCAPE` clause can make the intended escape character explicit.

### SQL Example — Topic 55

```sql
SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE 'R%';

SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE '_i%';

-- Example of searching for a literal percent sign in a text column:
SELECT employee_name
FROM operator_demo_employee
WHERE employee_name LIKE '%!%%' ESCAPE '!';
```

**Expected result / explanation**

The first pattern matches names beginning with R. The second matches names whose second character is i. The final query illustrates an explicit escape character; it returns no rows unless the searched text contains a literal percent sign.

## Topic 56: IS NULL Operator

`IS NULL` checks whether a column contains the SQL `NULL` marker, meaning the value is missing or unknown.

Do not write `column = NULL`; equality comparisons with `NULL` do not work as ordinary value comparisons.

### SQL Example — Topic 56

```sql
SELECT employee_name, bonus
FROM operator_demo_employee
WHERE bonus IS NULL;
```

**Expected result / explanation**

The result contains employees whose bonus has not been recorded as a value.

## Topic 57: IS NOT NULL Operator

`IS NOT NULL` checks whether a column contains a non-`NULL` value.

It is often used to find records with available information or to exclude missing values from a result.

### SQL Example — Topic 57

```sql
SELECT employee_name, bonus
FROM operator_demo_employee
WHERE bonus IS NOT NULL;
```

**Expected result / explanation**

Only employees with a recorded bonus value are returned.

## Topic 58: EXISTS Operator

`EXISTS` tests whether a subquery returns at least one row. It is commonly used with a correlated subquery, where the inner query refers to the current row of the outer query.

`NOT EXISTS` returns rows for which the subquery finds no matching row. The database can stop searching once it establishes whether a matching row exists.

### SQL Example — Topic 58

```sql
SELECT d.department_name
FROM (
    SELECT DISTINCT department AS department_name
    FROM operator_demo_employee
) AS d
WHERE EXISTS (
    SELECT 1
    FROM operator_demo_employee AS e
    WHERE e.department = d.department_name
      AND e.salary > 50000
);
```

**Expected result / explanation**

The query lists departments that have at least one employee earning more than 50,000. `EXISTS` checks whether the correlated subquery finds a qualifying employee.

## Topic 59: ANY Operator

`ANY` compares a value with the values returned by a subquery. The comparison is true if it is true for at least one value.

It is often used with operators such as `=`, `>`, `<`, `>=`, and `<=`. In MySQL, `SOME` is a synonym for `ANY`.

For example, `salary > ANY (subquery)` means the salary is greater than at least one value returned by the subquery.

### SQL Example — Topic 59

```sql
SELECT employee_name, salary
FROM operator_demo_employee
WHERE salary > ANY (
    SELECT salary
    FROM operator_demo_employee
    WHERE department = 'HR'
);
```

**Expected result / explanation**

The query returns employees whose salary is greater than at least one salary in the HR department. The subquery returns the HR salaries for comparison.

## Topic 60: ALL Operator

`ALL` compares a value with every value returned by a subquery. The comparison must be true for all returned values.

For example, `salary > ALL (subquery)` means the salary is greater than every value returned by the subquery.

`ANY` means at least one comparison succeeds; `ALL` means every comparison succeeds.

### SQL Example — Topic 60

```sql
SELECT employee_name, salary
FROM operator_demo_employee
WHERE salary > ALL (
    SELECT salary
    FROM operator_demo_employee
    WHERE department = 'HR'
);
```

**Expected result / explanation**

The query returns employees whose salary is greater than every salary in HR. If the subquery returns no rows, `ALL` comparisons evaluate as true; handle empty subquery cases carefully in application logic.

## Important Differences

### IN vs OR
- `IN ('IT', 'HR')` is concise for matching a list of values.
- `department = 'IT' OR department = 'HR'` expresses the same basic logic.

### BETWEEN vs Comparison Operators
- `salary BETWEEN 40000 AND 55000` includes both endpoints.
- The equivalent range condition is `salary >= 40000 AND salary <= 55000`.

### ANY vs ALL
- `> ANY` means greater than at least one value.
- `> ALL` means greater than every value.

### NULL Comparisons
- Use `IS NULL` and `IS NOT NULL`.
- Do not use `= NULL` or `!= NULL`.
