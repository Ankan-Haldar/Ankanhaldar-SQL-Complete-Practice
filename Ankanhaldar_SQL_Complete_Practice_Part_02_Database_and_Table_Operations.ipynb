## 18. CREATE DATABASE

**Definition:** `CREATE DATABASE` is a DDL command used to create a new database in MySQL.

**Syntax:**
```sql
CREATE DATABASE database_name;
```

**Important points:**
1. A database name should be meaningful.
2. The connected MySQL user must have permission to create databases.
3. `IF NOT EXISTS` prevents an error if a database with that name already exists.
4. Creating a database creates the database container; tables are created separately.

**Example:** Create a database named `ankan_sql_practice`.

````python
CREATE DATABASE ankan_sql_practice;
````

## Explanation

If the database does not already exist and you have permission, MySQL creates it.

**Safer alternative:** Use `IF NOT EXISTS` when you may run the setup statement more than once.

````python
CREATE DATABASE IF NOT EXISTS ankan_sql_practice;
````

## 19. DROP DATABASE

**Definition:** `DROP DATABASE` is a DDL command used to permanently remove a database, including its tables and stored data.

**Syntax:**
```sql
DROP DATABASE database_name;
```

**Important points:**
1. This operation is destructive.
2. Tables and data inside the database are removed.
3. You need the required privileges.
4. `IF EXISTS` avoids an error when the database does not exist.
5. Always confirm the database name before executing this command.

**Warning:** Never run `DROP DATABASE` on a production or important database unless deletion is intended and a verified backup exists.

````python
-- Destructive command: use only for a disposable practice database.
DROP DATABASE IF EXISTS temporary_practice_db;
````

## Explanation

The statement removes `temporary_practice_db` if it exists. It is deliberately separate from the main `ankan_sql_practice` database used in this notebook.

## 20. USE DATABASE

**Definition:** `USE` selects an existing database as the default database for the current MySQL session.

After selecting a database, you can refer to its tables without writing the database name before every table name.

**Syntax:**
```sql
USE database_name;
```

**Important points:**
1. The database must already exist.
2. The selection applies to the current connection/session.
3. You can change the selected database by running another `USE` statement.

````python
USE ankan_sql_practice;
````

## Explanation

After running this statement, unqualified table names refer to tables in `ankan_sql_practice` for the current session.

SHOW DATABASES;

**Uses:**
1. Check whether a database exists.
2. View databases available to the current user.
3. Verify that a newly created database appears in the list.

## 22. CREATE TABLE

**Definition:** `CREATE TABLE` is a DDL command used to create a new table inside a database. A table stores data in rows and columns.

**Syntax:**
```sql
CREATE TABLE table_name (
    column1 datatype constraints,
    column2 datatype constraints
);
```

**Important table elements:**
- **Column:** Defines the type of information stored.
- **Data type:** Defines the kind of value allowed.
- **Constraint:** Enforces a rule, such as `PRIMARY KEY` or `NOT NULL`.
- **Primary key:** Uniquely identifies each row.

**Example:** Create a `students` table with an ID, name, age and email.

````python
USE ankan_sql_practice;

CREATE TABLE students (
    student_id INT PRIMARY KEY AUTO_INCREMENT,
    student_name VARCHAR(100) NOT NULL,
    age INT,
    email VARCHAR(150) UNIQUE
);
````

## Explanation

- `student_id` is an integer primary key and automatically increases when a new row is inserted without an explicit ID.
- `student_name` must have a value because of `NOT NULL`.
- `age` stores an integer.
- `email` must be unique when it is provided. A nullable UNIQUE column can contain multiple NULL values in MySQL.

````python
-- Inspect the newly created table's structure.
SHOW CREATE TABLE students;
````

## 24. DESCRIBE TABLE

**Definition:** `DESCRIBE` (or `DESC`) is a MySQL statement used to display the structure of a table.

It shows information such as:
- Column names.
- Data types.
- Whether NULL values are allowed.
- Key information.
- Default values.
- Extra attributes, such as `auto_increment`.

**Syntax:**
```sql

```

**Note:** `DESCRIBE` is also commonly written as `DESC` in MySQL.

````python
DESCRIBE students;
````

## Explanation

The output should show `student_id`, `student_name`, `age` and `email`, along with their data types and constraints-related metadata.

## 25. ALTER TABLE

**Definition:** `ALTER TABLE` is a DDL command used to change the structure of an existing table without dropping and recreating the whole table.

**Common operations:**
1. Add a new column.
2. Modify a column's data type or definition.
3. Rename a column.
4. Drop a column.
5. Add or remove constraints.

**General syntax:**
```sql
ALTER TABLE table_name
operation;
```

The exact syntax for changing columns can vary between database systems. The examples below use MySQL syntax.

## 25.1 Add a Column

**Theory:** `ADD COLUMN` adds a new column to an existing table. Existing rows receive the specified default value, or NULL if allowed and no default is supplied.

````python
ALTER TABLE students
ADD COLUMN city VARCHAR(80);
````

## 25.2 Modify a Column

**Theory:** `MODIFY COLUMN` changes the definition of an existing column in MySQL, such as its data type or nullability. Make sure existing values are compatible with the new definition.

````python
ALTER TABLE students
MODIFY COLUMN student_name VARCHAR(150) NOT NULL;
````

## 25.3 Rename a Column

**Theory:** `RENAME COLUMN` changes a column's name while keeping the column's data and definition.

````python
ALTER TABLE students
RENAME COLUMN city TO home_city;
````

## 25.4 Drop a Column

**Theory:** `DROP COLUMN` removes a column and its stored values from the table. This is destructive for that column's data.

````python
ALTER TABLE students
DROP COLUMN home_city;
````

## 25.5 Add a Constraint

**Theory:** Constraints enforce rules on table data. For example, a CHECK constraint can require that an age value be non-negative.

````python
ALTER TABLE students
ADD CONSTRAINT chk_students_age
CHECK (age IS NULL OR age >= 0);
````

## Explanation

Each `ALTER TABLE` statement is a separate example. Run them in order if you want to follow the complete demonstration. After the final statement, the table has the updated column definition and an age check constraint.

## 26. RENAME TABLE

**Definition:** `RENAME TABLE` is a MySQL DDL statement used to change the name of an existing table.

**Syntax:**
```sql
RENAME TABLE old_table_name TO new_table_name;
```

**Important points:**
1. The table's rows and columns remain in the renamed table.
2. The old table name is no longer used.
3. You need the required privileges.
4. Update application queries that refer to the old name.

**Example:** Rename `students` to `college_students`.

````python
RENAME TABLE students TO college_students;
````

## Explanation

The table is now named `college_students`. Its existing data and columns remain in the table.

## 27. TRUNCATE TABLE

**Definition:** `TRUNCATE TABLE` is a DDL statement in MySQL that removes all rows from a table while keeping the table structure.

**Syntax:**
```sql
TRUNCATE TABLE table_name;
```

**Important points:**
1. All rows are removed; there is no `WHERE` clause.
2. The table and its columns remain.
3. In MySQL, truncating an InnoDB table generally resets its AUTO_INCREMENT counter.
4. It is not the same as deleting selected rows with `DELETE`.
5. It may be restricted when another table has a foreign key referencing the table.

**Warning:** This operation removes all records. Use it only on a practice table or when clearing all data is intended.

````python
-- Destructive: removes every row from this practice table.
TRUNCATE TABLE college_students;
````

## Explanation

The table `college_students` remains available, but all its rows are removed. Its table definition is not deleted.

## 28. DROP TABLE

**Definition:** `DROP TABLE` is a DDL statement used to permanently remove a table, including its structure and data.

**Syntax:**
```sql
DROP TABLE table_name;
```

**Important points:**
1. The table no longer exists after a successful DROP.
2. Its data and table definition are removed.
3. Use `IF EXISTS` to avoid an error if the table is absent.
4. Foreign key relationships may prevent a table from being dropped until dependencies are handled.

**Warning:** This is destructive. Confirm the table name and backup important data before running it.

````python
-- Destructive: removes the practice table completely.
DROP TABLE IF EXISTS college_students;
````

## Explanation

The `college_students` table is removed if it exists. Unlike `TRUNCATE`, `DROP TABLE` removes the table structure as well as its data.

## 29. INSERT INTO

**Definition:** `INSERT INTO` is a DML statement used to add new rows to a table.

**Common syntax:**
```sql
INSERT INTO table_name (column1, column2)
VALUES (value1, value2);
```

**Common ways to insert data:**
1. Insert one row.
2. Insert multiple rows in one statement.
3. Insert values for selected columns.
4. Insert rows using the result of a SELECT query.

**Best practice:** Specify column names explicitly. This makes the statement easier to read and safer if the table structure changes.

## 29.1 Insert One Row

**Theory:** Add one student record. The `student_id` is omitted because it is AUTO_INCREMENT.

````python
INSERT INTO students (student_name, age, email)
VALUES ('Ankan', 22, 'ankan@example.com');
````

## 29.2 Insert Multiple Rows

**Theory:** Multiple rows can be inserted in a single statement by separating each set of values with a comma.

````python
INSERT INTO students (student_name, age, email)
VALUES
    ('Rahul', 23, 'rahul@example.com'),
    ('Priya', 21, 'priya@example.com'),
    ('Amit', 24, 'amit@example.com');
````

## 29.3 Insert Selected Columns

**Theory:** You can provide values only for selected columns. Columns omitted from the INSERT receive their default value or NULL if allowed; an omitted required column without a default can cause an error.

````python
INSERT INTO students (student_name, age)
VALUES ('Riya', 20);
````

## 29.4 View Inserted Records

**Theory:** Use `SELECT` to check that the records were inserted successfully.

````python
SELECT * FROM students;
````

## Explanation

The first two INSERT statements add four records, and the third adds a record without an email. The exact auto-generated IDs depend on the table's AUTO_INCREMENT state. The final SELECT displays all rows.

## 30. UPDATE

**Definition:** `UPDATE` is a DML statement used to modify existing rows in a table.

**Syntax:**
```sql
UPDATE table_name
SET column_name = value
WHERE condition;
```

**Important points:**
1. `SET` specifies the new value.
2. `WHERE` selects which rows should be changed.
3. Without a WHERE clause, all rows in the table are updated.
4. You can update more than one column in a statement.
5. Always check the condition with a SELECT query before performing an important update.

**Warning:** An incorrect or missing WHERE condition can change every row.

## 30.1 Update One Column

**Theory:** Change the age of the student whose name is `Ankan`. This assumes there is one matching Ankan record in the practice table.

````python
UPDATE students
SET age = 23
WHERE student_name = 'Ankan';
````

## 30.2 Update Multiple Columns

**Theory:** Multiple columns can be updated in the same statement by separating assignments with commas.

````python
UPDATE students
SET age = 24,
    email = 'ankan.haldar@example.com'
WHERE student_name = 'Ankan';
````

## 30.3 Verify Updated Data

**Theory:** Use SELECT to verify the result of an update.

````python
SELECT *
FROM students
WHERE student_name = 'Ankan';
````

## Explanation

The first UPDATE changes the age to 23. The second UPDATE changes the age again to 24 and changes the email. The final SELECT displays the matching record(s). For safer updates, use a primary-key condition such as `WHERE student_id = 1` when you know the exact ID.

## 31. DELETE

**Definition:** `DELETE` is a DML statement used to remove one or more rows from a table.

**Syntax:**
```sql
DELETE FROM table_name
WHERE condition;
```

**Important points:**
1. `WHERE` selects the rows to delete.
2. Without `WHERE`, every row is deleted.
3. The table structure remains after DELETE.
4. DELETE can be used to remove selected records.
5. In a transaction using a transactional storage engine, changes may be rolled back before COMMIT.

**Warning:** Always verify the WHERE condition before deleting records.

## 31.1 Delete One Record

**Theory:** Delete the student whose email is `amit@example.com`.

````python
DELETE FROM students
WHERE email = 'amit@example.com';
````

## 31.2 Delete All Records

**Theory:** A DELETE statement without a WHERE clause removes every row but keeps the table structure.

**Warning:** This example is intentionally shown for learning. Do not run it unless you intend to empty the practice table.

````python
-- Destructive: deletes every row from students.
DELETE FROM students;
````

## 31.3 Verify Remaining Records

**Theory:** Use SELECT to check which records remain after a DELETE operation.

````python
SELECT * FROM students;
````

## Explanation

The first DELETE removes the matching Amit record. The second DELETE removes every row from `students`. Run only the DELETE example you intend to practise.

## REVISION: DELETE vs TRUNCATE vs DROP

| Feature | DELETE | TRUNCATE | DROP |
|---|---|---|---|
| SQL category | DML | DDL in MySQL | DDL |
| Removes rows | Yes | All rows | All rows with table |
| WHERE clause | Supported | Not supported | Not supported |
| Keeps table structure | Yes | Yes | No |
| Can target selected rows | Yes | No | No |
| Typical use | Remove selected records | Empty a table | Remove a table completely |

**Remember:**
- `DELETE` removes rows.
- `TRUNCATE` empties the table.
- `DROP` removes the table itself.

Exact transaction, trigger and foreign-key behaviour can differ by database system.
