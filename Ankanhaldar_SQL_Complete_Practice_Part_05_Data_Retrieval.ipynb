## Topic 61: SELECT Statement

`SELECT` retrieves data from one or more database tables. It is the most frequently used SQL statement for reading data.

Basic syntax:

```sql
SELECT column1, column2
FROM table_name;
```

Use `SELECT *` to retrieve all columns. In production queries, selecting only the columns needed is usually clearer and more efficient.

### SQL Example — Topic 61

````sql
SELECT employee_id, employee_name, department
FROM retrieval_demo_employee;

SELECT *
FROM retrieval_demo_employee;
````

**Expected result / explanation**

The first query retrieves only three selected columns. The second retrieves every column from the table.

## Topic 62: DISTINCT Keyword

`DISTINCT` removes duplicate rows from the query result.

- `SELECT DISTINCT department` returns each department once.
- `SELECT DISTINCT department, city` returns each unique department-city combination.
- `DISTINCT` applies to the complete selected column combination, not independently to each column.

### SQL Example — Topic 62

````sql
SELECT DISTINCT department
FROM retrieval_demo_employee;

SELECT DISTINCT department, city
FROM retrieval_demo_employee;
````

**Expected result / explanation**

The first query lists unique department names. The second lists unique department and city combinations.

## Topic 63: WHERE Clause

`WHERE` filters rows before they are returned. It accepts comparison, logical, membership, range, pattern, and NULL predicates.

General syntax:

```sql
SELECT columns
FROM table_name
WHERE condition;
```

Only rows for which the condition is true are included in the result.

### SQL Example — Topic 63

````sql
SELECT employee_name, department, salary
FROM retrieval_demo_employee
WHERE salary > 45000;

SELECT employee_name, city
FROM retrieval_demo_employee
WHERE city = 'Kolkata';
````

**Expected result / explanation**

The first query returns employees earning more than 45,000. The second returns employees located in Kolkata.

## Topic 64: ORDER BY Clause

`ORDER BY` sorts the result rows.

- `ASC` sorts in ascending order and is the default.
- `DESC` sorts in descending order.
- Multiple columns can be specified to define secondary sorting.

Sorting changes the displayed result order; it does not change the stored table data.

### SQL Example — Topic 64

````sql
SELECT employee_name, salary
FROM retrieval_demo_employee
ORDER BY salary ASC;

SELECT employee_name, department, salary
FROM retrieval_demo_employee
ORDER BY department ASC, salary DESC;
````

**Expected result / explanation**

The first query sorts salaries from lowest to highest. The second sorts by department, then by salary from highest to lowest within each department.

## Topic 65: GROUP BY Clause

`GROUP BY` combines rows that share the same value in one or more columns. It is commonly used with aggregate functions such as `COUNT()`, `SUM()`, `AVG()`, `MIN()`, and `MAX()`.

Every selected expression should either be grouped, aggregated, or otherwise valid under MySQL's grouping rules. Avoid relying on arbitrary non-grouped column values.

### SQL Example — Topic 65

````sql
SELECT department, COUNT(*) AS employee_count
FROM retrieval_demo_employee
GROUP BY department;

SELECT department, AVG(salary) AS average_salary
FROM retrieval_demo_employee
GROUP BY department;
````

**Expected result / explanation**

The first query counts employees in each department. The second calculates the average salary for each department.

## Topic 66: HAVING Clause

`HAVING` filters groups after `GROUP BY` has formed them. It is commonly used to filter aggregate results.

- `WHERE` filters individual rows before grouping.
- `HAVING` filters groups after grouping.

A query can use both clauses: `WHERE` first limits the input rows, and `HAVING` then filters the resulting groups.

### SQL Example — Topic 66

````sql
SELECT department, COUNT(*) AS employee_count
FROM retrieval_demo_employee
GROUP BY department
HAVING COUNT(*) >= 2;

SELECT department, AVG(salary) AS average_salary
FROM retrieval_demo_employee
GROUP BY department
HAVING AVG(salary) > 40000;
````

**Expected result / explanation**

The first query shows departments with at least two employees. The second shows departments whose average salary exceeds 40,000.

## Topic 67: LIMIT Clause

`LIMIT` restricts the maximum number of rows returned.

Syntax:

```sql
SELECT columns
FROM table_name
LIMIT number_of_rows;
```

It is often used for previews, pagination, and top-N queries. Without `ORDER BY`, the selected rows are not guaranteed to appear in a meaningful or predictable order.

### SQL Example — Topic 67

````sql
SELECT employee_id, employee_name, salary
FROM retrieval_demo_employee
ORDER BY salary DESC
LIMIT 3;
````

**Expected result / explanation**

The query returns the three highest-paid employees because the rows are sorted by salary in descending order before the limit is applied.

## Topic 68: OFFSET Clause

`OFFSET` skips a specified number of rows in a result set. It is commonly paired with `LIMIT` for pagination.

MySQL syntax:

```sql
LIMIT row_count OFFSET offset_value;
```

For page size 3:
- Page 1: `LIMIT 3 OFFSET 0`
- Page 2: `LIMIT 3 OFFSET 3`
- Page 3: `LIMIT 3 OFFSET 6`

Use a stable `ORDER BY` when implementing pagination.

### SQL Example — Topic 68

````sql
-- Page 1
SELECT employee_id, employee_name
FROM retrieval_demo_employee
ORDER BY employee_id
LIMIT 3 OFFSET 0;

-- Page 2
SELECT employee_id, employee_name
FROM retrieval_demo_employee
ORDER BY employee_id
LIMIT 3 OFFSET 3;
````

**Expected result / explanation**

The first query returns rows 1–3 in employee ID order. The second skips the first three and returns the next three.

## Topic 69: Column Aliases

A **column alias** gives a temporary name to a column or calculated expression in a query result.

Use `AS` for readability. The alias affects the output heading; it does not rename the actual table column.

### SQL Example — Topic 69

````sql
SELECT
    employee_name AS name,
    salary AS monthly_salary,
    salary * 12 AS annual_salary
FROM retrieval_demo_employee;
````

**Expected result / explanation**

The output headings are `name`, `monthly_salary`, and `annual_salary`. The original table schema remains unchanged.

## Topic 70: Table Aliases

A **table alias** gives a shorter temporary name to a table within a query. It is especially useful when joining tables or when the same table is referenced more than once.

The alias is valid only within that query. In MySQL, the `AS` keyword is optional for table aliases, but using it can improve readability.

### SQL Example — Topic 70

````sql
SELECT
    e.employee_id,
    e.employee_name,
    e.department
FROM retrieval_demo_employee AS e
WHERE e.salary >= 50000;
````

**Expected result / explanation**

`e` is the table alias. It lets the query refer to `retrieval_demo_employee` using a shorter name.

## Topic 71: CASE Expression

`CASE` adds conditional logic to a SQL query. It evaluates conditions in order and returns the result for the first matching `WHEN`. If no condition matches, `ELSE` is used; without `ELSE`, the result is `NULL`.

General form:

```sql
CASE
    WHEN condition1 THEN result1
    WHEN condition2 THEN result2
    ELSE default_result
END
```

### SQL Example — Topic 71

````sql
SELECT
    employee_name,
    salary,
    CASE
        WHEN salary >= 55000 THEN 'High Salary'
        WHEN salary >= 40000 THEN 'Medium Salary'
        ELSE 'Entry Salary'
    END AS salary_category
FROM retrieval_demo_employee;
````

**Expected result / explanation**

Each employee is assigned a salary category according to the first matching condition. The order of `WHEN` conditions matters.

## Topic 72: IF() Function

MySQL's `IF()` function returns one value when a condition is true and another when it is false.

Syntax:

```sql
IF(condition, value_if_true, value_if_false)
```

`IF()` is a MySQL-specific function. `CASE` is standard SQL and is more flexible for multiple conditions.

### SQL Example — Topic 72

````sql
SELECT
    employee_name,
    salary,
    IF(salary >= 45000, 'Eligible', 'Not Eligible') AS salary_check
FROM retrieval_demo_employee;
````

**Expected result / explanation**

The output shows `Eligible` for salaries of 45,000 or more and `Not Eligible` otherwise.

## Topic 73: DISTINCT vs GROUP BY

Both `DISTINCT` and `GROUP BY` can produce one row per unique value combination, but their main purposes differ.

**DISTINCT**
- Removes duplicate rows from the selected result.
- Commonly used when only unique values are needed.

**GROUP BY**
- Forms groups of rows.
- Commonly used with aggregate functions to calculate counts, totals, averages, and other summaries.

A simple `GROUP BY` without aggregates can produce results similar to `DISTINCT`, but the intent is different.

### SQL Example — Topic 73

````sql
-- DISTINCT: list each department once
SELECT DISTINCT department
FROM retrieval_demo_employee;

-- GROUP BY: count employees in each department
SELECT department, COUNT(*) AS employee_count
FROM retrieval_demo_employee
GROUP BY department;
````

**Expected result / explanation**

Use `DISTINCT` to remove duplicate result rows. Use `GROUP BY` when you need to summarize records by group.

## Topic 74: WHERE vs HAVING

`WHERE` and `HAVING` both filter query results, but they operate at different stages.

| WHERE | HAVING |
|---|---|
| Filters individual rows | Filters groups |
| Applied before grouping | Applied after grouping |
| Commonly filters ordinary columns | Commonly filters aggregate results |
| Does not directly filter aggregate results at the same query level | Can filter `COUNT()`, `SUM()`, `AVG()`, etc. |

A query can use both clauses.

### SQL Example — Topic 74

````sql
SELECT
    department,
    COUNT(*) AS employee_count,
    AVG(salary) AS average_salary
FROM retrieval_demo_employee
WHERE employment_status = 'Active'
GROUP BY department
HAVING COUNT(*) >= 1;
````

**Expected result / explanation**

`WHERE` first excludes inactive employees. `GROUP BY` then forms department groups. `HAVING` keeps groups with at least one active employee.

## Topic 75: DELETE vs TRUNCATE vs DROP

These statements remove data or database objects, but their effects differ.

| DELETE | TRUNCATE | DROP |
|---|---|---|
| Removes selected rows (or all rows without `WHERE`) | Removes all rows from a table | Removes the table object itself |
| Supports `WHERE` | Does not support `WHERE` | Removes both data and structure |
| Table structure remains | Table structure remains | Table structure is removed |
| DML statement | DDL statement in MySQL | DDL statement |
| Can delete selected rows | Empties the whole table | Deletes the table |

**Important:** `TRUNCATE` and `DROP` are destructive. MySQL DDL statements cause implicit commits in common configurations, so do not assume they can be rolled back like ordinary transactional DML. Always verify the target before running them.

## SQL SELECT Query Processing Order

A simplified logical processing order for a typical query is:

1. `FROM` — identify source tables.
2. `WHERE` — filter source rows.
3. `GROUP BY` — form groups.
4. `HAVING` — filter groups.
5. `SELECT` — calculate selected expressions.
6. `DISTINCT` — remove duplicate result rows.
7. `ORDER BY` — sort the result.
8. `LIMIT` / `OFFSET` — restrict and skip result rows.

This is the logical processing order, not necessarily the physical execution plan chosen by the database optimizer.
